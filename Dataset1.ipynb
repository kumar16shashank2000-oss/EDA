```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df= pd.read_csv("googleplaystore.csv")
```

```python
## Count numerical and categorical columns

nums_cols= df.select_dtypes(include='number').columns
cate_cols= df.select_dtypes(include='').colums
```

```python
df.duplicated().sum()
```

```python
# Remove duplicate records

df.drop_duplicates(inplace=True)
```

```python
# Step1: Inspect the column Installs

df['Installs'].head()
```

```python
# Step2: Check for unique values

df['Installs'].unique()
```

```python
# Step3: Drop free value

df.drop(df[df['Installs'] == 'Free'].index,inplace=True)
```

```python
# removing "+" , "," from the data

df['Installs'] = df['Installs'].str.replace(',','', regex=False)            # regex= regular expression
df['Installs'] = df['Installs'].str.replace('+','', regex=False)
```

```python
df['Installs'].head()
```

```python

df['Installs'] = df['Installs'].astype(int)
```

```python
# Step1; Inspect the column Price

df['Price'].head()
```

```python
df['Price'].unique()
```

```python
df['Price']= df['Price'].str.replace('$',' ', regex=False)
df['Price']= df['Price'].astype(float)
```

```python
# Clean the "Review " column

df['Reviews']= df['Reviews'].astype(int)
```

```python
df['Size'].unique()
```

```python
df['Size']= df['Size'].str.replace('M','',regex=False)
df['Size']= df['Size'].str.replace('k','',regex=False)
```

```python
df['Size']= df['Size'].replace('Varies with device',np.nan)

df['Size']= df['Size'].astype(float)
```

```python
df['Last Updated'].head()
```

```python
# 

df['Last Updated']= pd.to_datetime(df['Last Updated'])
```
